```python
#====================================
# Calculate Total Engagement
#====================================
```

```python
post["total_engagement"] = (
    post["likes"] + post["comments"]
)

post[
    ["id", "likes", "comments", "total_engagement"]
].head()
```

```python
#=======================================
# Calculate Followers
#=======================================
```

```python
follower_counts = (
    follows.groupby("followee ")
    ["follower"]
    .nunique()
    .reset_index()
)
```

```python
follower_counts.columns = [
    "user ID",
    "followers"
]
```

```python
post = post.merge(
    follower_counts,
    on="user ID",
    how="left"
)
```

```python
post["followers"] = post["followers"].fillna(0)
```

```python
#======================================
# Calculate Likes Per Follower
#======================================
```

```python
post["likes_per_follower"] = np.where(
    post["followers"] > 0,
    post["likes"] / post["followers"],
    np.nan
)
```

```python
#======================================
# Calculate Comments Per Follower
#======================================
```

```python
post["comments_per_follower"] = np.where(
    post["followers"] > 0,
    post["comments"] / post["followers"],
    np.nan
)
```

```python
#=============================================
# Calculate Overall Engagement Per Follower
#=============================================
```

```python
post["engagement_per_follower"] = np.where(
    post["followers"] > 0,
    post["total_engagement"] / post["followers"],
    np.nan
)
```

```python
#========================================
# Analyze Content Types
#========================================
```

```python
post["photo type"].value_counts()
```

```python
content_analysis = (
    post.groupby("photo type")
    .agg(
        posts=("id", "count"),
        avg_likes=("likes", "mean"),
        avg_comments=("comments", "mean"),
        avg_engagement=("total_engagement", "mean"),
        avg_engagement_per_follower=(
            "engagement_per_follower", "mean"
        )
    )
)
```

```python
content_analysis
```

```python
#==============================================
# Create Content-Type Visualization
#==============================================
```

```python
content_analysis["avg_engagement"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Engagement by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Average Engagement")
plt.xticks(rotation=0)
plt.show()
```

```python
#=========================================
# Analyze Hashtags
#=========================================
```

```python
tag_data = photo_tags.merge(
    tags,
    left_on="tag ID",
    right_on="id",
    how="left"
)
```

```python
tag_data = tag_data.merge(
    post[
        ["id", "likes", "comments", "total_engagement",
         "engagement_per_follower"]
    ],
    left_on="photo",
    right_on="id",
    how="left"
)
```

```python
#========================================
# Calculate Hashtag Performance
#========================================
```

```python
hashtag_analysis = (
    tag_data.groupby("tag text")
    .agg(
        posts=("photo", "nunique"),
        avg_likes=("likes", "mean"),
        avg_comments=("comments", "mean"),
        avg_engagement=("total_engagement", "mean"),
        avg_engagement_per_follower=(
            "engagement_per_follower", "mean"
        )
    )
    .sort_values(
        "avg_engagement",
        ascending=False
    )
)
```

```python
hashtag_analysis.head(10)
```
